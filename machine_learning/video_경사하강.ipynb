 # 2-1. `3*time`

```
3 * time
```

 앞에서 `time`이:

```
[0, 1, 2, 3, ..., 19]
```

 였죠.

 여기에 각각 3을 곱합니다.

```
3 × [0, 1, 2, 3, ..., 19]
```

 따라서:

```
[0, 3, 6, 9, 12, ..., 57]
```

 이 됩니다.

 수학적으로는:

 $$
3t
$$

 입니다.

---

 # 2-2. `3*time - 2`

 이제 각각의 값에서 2를 뺍니다.

```
3*time - 2
```

 따라서:

```
time = 0 → 3×0 - 2 = -2
time = 1 → 3×1 - 2 =  1
time = 2 → 3×2 - 2 =  4
time = 3 → 3×3 - 2 =  7
...
```

 즉 기본적인 가격은 다음과 같은 직선 형태입니다.

 $$
price = 3t - 2
$$

 여기서 중요한 부분이 있습니다.

 ### `3`은 기울기입니다.

```
price = 3 × time - 2
        ↑
      기울기
```

 시간이 1 증가할 때마다 가격은 평균적으로 **3만큼 증가**합니다.

 ### `-2`는 절편입니다.

```
price = 3 × time - 2
                   ↑
                 절편
```

 시간이 0일 때:

 $$
price = 3(0)-2=-2
$$

 입니다.

 따라서 기본적으로 만들어지는 데이터는 **기울기 3, 절편 -2인 직선**을 따릅니다.

---

 # 2-3. `torch.randn(len(time))`

 이 부분이 조금 중요합니다.

```
torch.randn(len(time))
```

 `randn()`은 **정규분포(normal distribution)를 따르는 랜덤한 숫자**를 만들어냅니다.

 예를 들어 실행할 때마다 대략 이런 값이 나올 수 있습니다.

```
tensor([
    0.21,
   -0.73,
    1.15,
    0.08,
   -1.42,
    ...
])
```

 중요한 특징은 평균이 대략 **0**이고 표준편차가 **1**인 정규분포에서 숫자를 뽑는다는 것입니다.

 즉:

```
torch.randn(...)
```

 은 쉽게 말하면:

 > "평균 0 근처에서 랜덤한 오차를 만들어줘."

 라고 이해하면 됩니다.

---

 ## 그런데 왜 `len(time)`일까요?

```
len(time)
```

 을 먼저 보면:

```
time = [0, 1, 2, ..., 19]
```

 이므로 데이터가 **20개**입니다.

 따라서:

```
len(time)
```

 은

```
20
```

 입니다.

 그러므로:

```
torch.randn(len(time))
```

 은 결국:

```
torch.randn(20)
```

 과 같습니다.

 즉 랜덤 숫자 **20개**를 생성합니다.

 예를 들어:

```
[ 0.4, -0.8, 0.1, 1.3, -0.5, ...]
```

 처럼 됩니다.

---

 # 2-4. `torch.randn(len(time)) * 3`

 이제 랜덤 숫자에 3을 곱합니다.

```
torch.randn(len(time)) * 3
```

 예를 들어 랜덤 숫자가:

```

 이것을 **노이즈(noise), 즉 오차**라고 생각하면 이해하기 쉽습니다.

 # 2-5\. 전체를 합쳐보겠습니다

 다시 원래 코드로 돌아가겠습니다.

```
price = 3*time - 2 + torch.randn(len(time)) * 3
```

 수학적으로 표현하면:

 $$
price = 3t - 2 + noise
$$

 입니다.

 조금 더 정확하게 쓰면:

 $$
price_i = 3t_i - 2 + \epsilon_i
$$

 입니다.

 여기서:

 - $t_i$ = 시간
- $3t_i-2$ = 우리가 의도적으로 만든 기본적인 가격
- $\epsilon_i$ = 랜덤 오차(noise)
- `price_i` = 최종적으로 만들어진 가격

 입니다.

 # 왜 굳이 랜덤 오차를 넣었을까요?

```

```
price
  |
55|                 ●
  |              ●
40|           ●
  |        ●
25|     ●
  |  ●
10|
  +---------------------- time
```

 그런데 실제 현실의 데이터는 이렇게 정확하게 직선으로 나오지 않습니다.

```
시간 증가 → 가격 증가
```

 하면서도 매번 정확히 3씩 증가하지는 않습니다.

 그래서:

```
+ torch.randn(len(time)) * 3
```

 를 추가하여 **현실적인 데이터처럼 약간의 흔들림을 만들어준 것**입니다.

 ## 이 예제가 머신러닝에서 중요한 이유

 이 데이터는 사실 머신러닝에서 아주 중요한 형태입니다.

 우리가 데이터를 만들 때는 이미 정답을 알고 있습니다.

```
price = 3*time - 2 + noise
```

 즉 실제 데이터가 만들어진 규칙은 대략:

 $$
y = 3x - 2
$$

 입니다.

 그런데 머신러닝에서는 이 **3과 -2를 우리가 직접 알려주지 않고**, 데이터만 보여준 다음 모델이 스스로 찾아내도록 합니다.

 예를 들어 모델이 다음과 같은 식을 학습한다고 생각해 보세요.

 $$
y = ax + b
$$

 처음에는:

```
a = 아무 값
b = 아무 값
```

 일 수 있습니다.

 그런데 데이터를 계속 학습하면서:

```
a ≈ 3
b ≈ -2
```

 에 가까워지도록 만드는 것이 **선형 회귀의 핵심 아이디어**입니다.

 따라서 지금 코드에서 특히 기억해두면 좋은 것은:

```
price = 3*time - 2 + torch.randn(len(time)) * 3
```

 가 단순한 숫자 생성 코드가 아니라,

 > **“정답이 대략 `y = 3x - 2`인 데이터를 만들고, 여기에 랜덤한 오차를 추가해서 실제 데이터처럼 만든다.”**

 라는 의미라는 것입니다.

 그리고 바로 다음 단계에서 이 데이터를 이용해 **`a`와 `b`를 머신러닝이 어떻게 찾아가는지**를 보면 `loss`, `gradient`, `optimizer`, `fit` 같은 개념이 훨씬 쉽게 연결됩니다.

````python
def f(t, params):
    a, b = params
    return a * t * b
````

* 최종 시각화 동작: 함수 내부 로직에 따라 화면에는 파란색 점(실제 time별 price 데이터)이 먼저 그려지고, 그 위에 현재 매개변수로 예측한 값들이 빨간색 점으로 겹쳐서 나타납니다.

## 3. 전체 흐름 요약 (지금까지 질문하신 세 코드의 연결)
지금까지 질문하신 세 가지 코드를 순서대로 엮으면 완전한 머신러닝 초기화 및 확인 단계가 됩니다.

   1. params = torch.randn(2).requires_grad_() $\rightarrow$ 예측에 쓸 가중치(파라미터)를 무작위로 초기화하고 미분 준비를 합니다.
   2. preds = f(time, params) $\rightarrow$ 이 무작위 가중치와 시간 데이터를 가지고 첫 번째 예측값(preds)을 계산합니다.
   3. show_preds(preds) $\rightarrow$ 무작위 상태의 가중치로 만든 예측(빨간 점)이 실제 데이터(파란 점)와 얼마나 동떨어져 있는지 그래프로 확인합니다. (초기 상태이므로 보통 실제 데이터와 잘 맞지 않고 엉뚱한 곳에 빨간 점이 찍히게 됩니다.)

이 초기 예측 이후에 예측 오차(Loss)를 줄이기 위해 가중치를 수정하는 '학습(Training)' 과정을 코드로 구현하는 방법이 궁금하신가요? 원하신다면 경사하강법(Gradient Descent)을 통해 빨간 점이 파란 점으로 이동하는 최적화 코드를 이어서 설명해 드릴 수 있습니다.



3. calculate the loss

````python
loss = mse(preds, price) # mean squared loss
loss
````

제시해주신 파이썬 코드는 머신러닝 및 딥러닝 모델의 학습 과정에서 예측 값과 실제 값이 얼마나 차이 나는지 계산하는 코드입니다. 인공지능이 모델의 예측 오류를 측정하기 위해 가장 흔하게 사용하는 방식 중 하나입니다.
코드의 구성 요소를 한 부분씩 자세히 해석해 드리겠습니다.
------------------------------
## 1. 코드 요소별 상세 해석

* 의미: 계산된 오차(손실) 값을 loss라는 변수에 저장한다는 뜻입니다.
   * 역할: 모델은 이 loss 값을 바탕으로 "내가 지금 얼마나 못 맞추고 있는지"를 파악하고, 이 값을 0에 가깝게 줄이는 방향으로 학습(가중치 업데이트)을 진행합니다.
* mse(...)
* 의미: Mean Squared Error(평균 제곱 오차)의 약자로, 오차를 계산하는 함수입니다.
   * 역할: 예측값과 실제값의 차이를 구한 뒤, 그 값을 제곱하여 평균을 내는 수학적 연산을 수행합니다. 음수 오차를 방지하고 큰 오차에 더 큰 벌점(penalty)을 주는 특성이 있습니다.
* preds
* 의미: Predictions(예측값)의 줄임말입니다.
   * 역할: 머신러닝 모델이 학습을 통해 나름대로 계산해서 내놓은 예측 결과물입니다. (예: "이 집의 가격은 5억 원일 것 같다.")

* 의미: 실제 정답(Target/Label) 데이터입니다. 여기서는 '가격' 데이터에 해당합니다.

   * 역할: 코드가 실행될 때는 아무런 영향을 주지 않으며, 이 코드가 "평균 제곱 오차를 구하는 손실 함수"임을 사람에게 설명해 주는 역할을 합니다.

## 2. 작동 방식 예시 (수학적 의미)
만약 데이터가 1개 있고, 모델의 예측과 실제 값이 다음과 같다면 내부적으로 어떻게 계산되는지 살펴보겠습니다.

   1. 차이 구하기 (오차): preds (5억) - price (4억 5천만) = +5천만
   2. 제곱하기: 오차를 제곱합니다. (음수가 나와도 제곱하면 양수가 되어 방향성에 상관없이 거리만 측정할 수 있게 됩니다.)
   3. 평균 내기: 데이터가 여러 개라면 이 제곱된 오차들을 모두 더한 뒤 데이터 개수로 나눕니다.

최종적으로 나온 이 loss 값이 크면 모델의 예측 성능이 나쁜 것이고, 0에 가까울수록 모델이 완벽하게 예측하고 있음을 뜻합니다.
이 코드가 사용되는 전체적인 학습 흐름(역전파, 최적화)이 궁금하시거나, PyTorch나 TensorFlow 같은 특정 라이브러리에서의 구현 방식이 궁금하시면 언제든 말씀해 주세요!



4. calculate the gradient

````python
loss.backward()
params.grad
````

제시해주신 코드는 파이토치(PyTorch) 같은 딥러닝 프레임워크에서 모델을 학습시키기 위해 오차를 역방향으로 전달하고(역전파), 각 가중치(파라미터)가 수정되어야 할 방향과 크기(기울기)를 계산하는 핵심 단계입니다.
앞서 계산한 loss를 바탕으로 모델이 '반성'하고 '공부'하는 과정에 해당합니다. 코드의 구성 요소를 하나씩 자세히 해석해 드리겠습니다.
------------------------------
## 1. 코드 요소별 상세 해석

* loss.backward()
* 의미: 오차(Loss)로부터 시작하여 모델의 처음 방향으로 거꾸로 타고 올라가며 역전파(Backpropagation)를 수행하라는 명령입니다.
   * 역할: 최종 오차에 각 파라미터(가중치와 편향)가 얼마나 기여했는지 그 영향력을 계산합니다. 수학적으로는 미분의 '체인 룰(Chain Rule, 연쇄 법칙)'을 이용해 최종 손실에 대한 각 파라미터의 미분값(기울기)을 계산하는 작업을 합니다.
* params.grad
* 의미: 모델 파라미터(params)들의 기울기(Gradient, 그라디언트) 저장소입니다.
   * 역할: loss.backward()가 실행되고 나면, 계산된 미분값들이 이 grad라는 변수 안에 자동으로 저장됩니다. 이 값은 "오차를 줄이기 위해 이 가중치를 어느 방향으로, 얼마나 많이 움직여야 하는가?"를 알려주는 나침반 역할을 합니다.

------------------------------
## 2. 비유를 통한 쉽게 이해하기 (등산 비유)
이 과정은 안개가 자욱한 산속(손실 함수 그래프)에서 가장 낮은 골짜기(오차가 0인 지점)를 찾아 내려가는 과정과 같습니다.

   1. loss (현재 위치의 높이): 현재 내 위치가 얼마나 높은지(오차가 얼마나 큰지) 확인합니다.
   2. loss.backward() (주변 경사 조사): 발을 사방으로 뻗어보며 내가 서 있는 자리의 경사도(기울기)를 측정합니다.
   3. params.grad (내려갈 방향 확인): "앞쪽으로 3도 경사, 오른쪽으로 5도 경사"처럼 어느 쪽이 내려가는 방향인지 알아낸 경사값의 결과물입니다.

이 params.grad 값을 구하고 나면, 다음 단계에서는 최적화 알고리즘(Optimizer)을 사용해 params 값을 경사가 낮은 쪽으로 한 걸음 이동시키게 됩니다(예: optimizer.step()).
이제 기울기(grad)를 구했으니, 이를 실제로 모델에 반영하여 가중치를 업데이트하는 최적화(Optimizer/경사하강법) 단계로 넘어갈 수 있습니다.

* optimizer.step()이나 optimizer.zero_grad() 같은 다음 학습 코드가 어떻게 연결되는지 궁금하신가요?
* 아니면 이 과정이 진행되는 전체 인공지능 학습 루프(Train Loop) 코드가 보고 싶으신가요?




5. step the weights (parameters)

````python
lr = 1e-4

params.data -= lr * params.grad.data
params.grad = None
````

제시해주신 코드는 딥러닝 학습의 최종 단계인 최적화(Optimization, 경사하강법) 과정입니다. 앞서 loss.backward()를 통해 구한 기울기(grad)를 바탕으로, 모델의 가중치(파라미터)를 실제로 업데이트하고 다음 학습을 위해 초기화하는 작업을 수행합니다.
이 코드는 파이토치(PyTorch)의 내부 최적화 기반 알고리즘(예: optimizer.step()과 optimizer.zero_grad())이 실제로 수행하는 원리를 수동으로 구현한 것입니다. 각 줄을 자세히 해석해 드리겠습니다.
------------------------------
## 1. 코드 요소별 상세 해석## 1) lr = 1e-4

* 의미: 학습률(Learning Rate, $lr$)을 0.0001($10^{-4}$)로 설정합니다.
* 역할: 가중치를 한 번에 얼마나 크게 업데이트할지 결정하는 보폭입니다.
* 값이 너무 크면 정답을 지나쳐버릴 수 있고, 너무 작으면 학습이 지나치게 오래 걸립니다. 1e-4는 딥러닝에서 매우 자주 쓰이는 기본 학습률 중 하나입니다.

## 2) params.data -= lr * params.grad.data

* 의미: 기존 가중치(params.data)에서 [학습률 $\times$ 기울기] 값을 빼서 가중치를 업데이트합니다.
* 역할: 오차를 줄이는 방향으로 모델의 파라미터를 한 걸음 이동시키는 경사하강법(Gradient Descent)의 핵심 공식입니다.
* 왜 빼나요? 기울기(grad) 방향은 오차가 '커지는' 방향입니다. 우리는 오차를 '줄여야' 하므로, 기울기의 반대 방향(-)으로 가중치를 이동시켜야 합니다.
   * .data를 쓰는 이유: 파이토치에서 연산 그래프(추적 기능)에 영향을 주지 않고, 순수하게 데이터 숫자만 빠르게 수정하기 위해 임시로 접근하는 방식입니다.

## 3) params.grad = None

* 의미: 계산에 사용했던 가중치의 기울기(grad) 메모리를 비워줍니다(초기화).
* 역할: 파이토치는 기본적으로 loss.backward()를 호출할 때마다 기존 기울기에 새로운 기울기를 더하는(누적하는) 성질이 있습니다.
* 만약 이 줄을 생략하면, 다음 데이터(Batch)를 학습할 때 이전 오차의 기억이 그대로 더해져서 완전히 잘못된 방향으로 학습하게 됩니다. 따라서 한 번 가중치를 바꾼 후에는 반드시 깨끗하게 비워야 합니다. (라이브러리에서는 optimizer.zero_grad()가 이 역할을 합니다.)

------------------------------
## 2. 가중치 업데이트 과정 시각화 (수학적 원리)
만약 어떤 파라미터의 현재 값이 5.0이고, 기울기(grad)가 2.0으로 계산되었다면 가중치는 다음과 같이 바뀝니다.
$$\text{새로운 가중치} = 5.0 - (0.0001 \times 2.0) = 4.9998$$
오차를 줄이기 위해 가중치 값이 5.0에서 4.9998로 아주 미세하게 정답을 향해 조정된 것입니다. 이 작업을 수만 번 반복하면서 모델이 점점 똑똑해지게 됩니다.

````python
preds = f(time, params)
mse(preds, price)
````

제시해주신 코드는 머신러닝의 가장 첫 단계인 '예측(순전파)'과 '오차 계산'이 이루어지는 과정입니다.
시간(time) 데이터를 바탕으로 가중치(params)를 가진 함수 f를 통과시켜 예측값을 구하고, 이 값이 실제 가격(price)과 얼마나 차이가 나는지 계산하는 구조입니다. 구성 요소를 하나씩 자세히 해석해 드리겠습니다.
------------------------------
## 1. 코드 요소별 상세 해석## 1) preds = f(time, params)

* f
* 의미: 예측을 수행하는 모델(함수, 모델 구조)을 뜻합니다.
   * 역할: 입력값과 가중치를 조합하여 결과를 만들어내는 수학적 공식이나 신경망 층(Layer)에 해당합니다.
* time
* 의미: 모델에 집어넣는 입력 데이터(Feature)입니다. 여기서는 '시간' 정보입니다.
   * 역할: 원인에 해당하는 변수입니다. 예를 들어 "주식 거래가 시작된 지 10분이 지났다", "집을 지은 지 5년이 되었다"와 같은 정보입니다.
* params
* 의미: 모델이 학습해야 하는 가중치와 편향(Parameters/Weights)입니다.

* preds =
* 의미: 함수 f에 time과 params를 넣어서 계산되어 나온 예측 결과물(Predictions)을 preds라는 변수에 저장합니다. (예: "10분 뒤 이 주식의 예측 가격은 5,000원이다.")

## 2) mse(preds, price)

* 의미: 방금 구한 예측값(preds)과 실제 정답 가격(price) 사이의 평균 제곱 오차(Mean Squared Error)를 계산합니다.
* 역할: 이 값이 첫 질문에서 보셨던 loss = mse(preds, price)의 그 연산입니다. 이 줄이 실행되면 모델이 방금 예측한 값이 실제 값과 얼마나 틀렸는지 '점수(오차)'로 환산되어 나오게 됩니다.

## 🔄 전체 그림 완성하기 (학습의 한 바퀴)
질문해주신 코드는 전체 인공지능 학습 흐름의 출발점에 해당합니다. 지금까지 물어보신 모든 코드를 순서대로 정렬하면, 딥러닝 모델이 한 번 학습하는 전체 메커니즘이 완벽하게 연결됩니다.

# [1단계] 예측하기 (순전파 - Forward Pass)# "시간과 가중치를 가지고 가격을 예측해봐"preds = f(time, params)
# [2단계] 오차 계산하기 (손실 계산 - Loss Computation)# "실제 가격이랑 비교해서 얼마나 틀렸는지 점수를 내봐"loss = mse(preds, price)
# [3단계] 원인 분석하기 (역전파 - Backward Pass)# "틀린 오차를 바탕으로 각 가중치(params)가 얼마나 잘못 기여했는지 계산해봐"
loss.backward()
# [4단계] 가중치 수정 및 초기화 (최적화 - Optimization)# "기울기(grad) 반대 방향으로 가중치를 조금 수정하고, 다음을 위해 메모리를 비워줘"lr = 1e-4
params.data -= lr * params.grad.data
params.grad = None

인공지능은 이 1~4단계를 수천, 수만 번 반복하면서 params를 최적의 값으로 찾아가게 됩니다.
이제 전체적인 원리를 모두 파악하셨습니다! 혹시 이 원리를 바탕으로:

* 실제로 작동하는 가상의 간단한 함수 f와 데이터를 넣어서 전체 코드가 돌아가는 예시를 보고 싶으신가요?
* 아니면 파이토치(PyTorch)의 실제 인공지능 모델 구조(nn.Module)로는 어떻게 작성하는지 궁금하신가요?




6. repeat

````python
def apply_step(params, prn=True):
  # step 2~5
  preds = f(time, params)
  loss = mse(preds, price)
  loss.backward()
  params.data -= lr * params.grad.data
  params.grad = None
  if prn: print(loss.item())
  return preds
````

제시해주신 코드는 지금까지 우리가 하나씩 짚어보았던 딥러닝 학습의 핵심 4단계(예측, 오차 계산, 역전파, 가중치 수정)를 하나의 함수로 묶어 자동화한 코드입니다.
이 함수를 한 번 호출할 때마다 모델은 데이터를 한 번 학습(1 Step)하게 됩니다. 각 줄의 의미를 자세히 해석해 드리겠습니다.
------------------------------
## 1. 코드 요소별 상세 해석## 1) def apply_step(params, prn=True):

* 의미: apply_step이라는 이름의 함수를 정의합니다. 이 함수는 두 개의 입력값을 받습니다.
* params: 업데이트할 모델의 가중치(파라미터)입니다.
   * prn=True: '출력(print) 여부'를 결정하는 옵션(기본값은 True)입니다. 함수를 쓸 때 이 값을 따로 지정하지 않으면 자동으로 화면에 오차를 출력합니다.

## 2) # step 2~5

* 의미: 개발자가 작성한 주석입니다. 이 함수 내부의 작업들이 학습 과정의 2단계부터 5단계에 해당함을 메모해 둔 것입니다.

## 3) 학습 핵심 루프 (내부 코드)

* preds = f(time, params)
* 순전파(Forward): 현재 가중치(params)와 입력값(time)으로 새로운 예측값(preds)을 계산합니다.
* loss = mse(preds, price)
* 오차 계산(Loss): 예측값과 실제 정답(price)을 비교하여 오차(점수)를 구합니다.
* loss.backward()
* 역전파(Backward): 오차를 바탕으로 각 가중치가 수정되어야 할 방향과 크기(기울기, grad)를 계산합니다.
* params.data -= lr * params.grad.data
* 가중치 수정(Update): 계산된 기울기 방향의 반대로 가중치를 학습률(lr) 보폭만큼 이동시킵니다.
* params.grad = None
* 기울기 초기화(Reset): 다음 번 학습 때 기울기가 누적되어 꼬이지 않도록 메모리를 깨끗하게 비웁니다.

## 4) if prn: print(loss.item())

* 의미: 만약 prn 변수가 True라면, 현재 학습 단계의 오차(Loss) 값을 화면에 출력하라는 뜻입니다.
* .item()의 역할: 파이토치의 loss는 단순한 숫자가 아니라 연산 그래프가 포함된 텐서(Tensor) 객체입니다. .item()을 붙여주면 순수한 파이썬 숫자(Float)만 쏙 뽑아내어 출력하기 편하고 메모리도 아낄 수 있습니다. 학습이 잘되고 있다면 함수를 반복 호출할 때마다 이 출력되는 숫자가 점점 줄어들게 됩니다.

## 5) return preds

* 의미: 함수가 끝날 때 최종적으로 계산된 예측값(preds)을 반환합니다.
* 역할: 이 함수를 호출한 바깥 코드에서 "이번 단계에서는 모델이 어떻게 예측했는지" 그 결과물을 받아서 시각화하거나 다른 용도로 활용할 수 있게 해줍니다.

````python
def f(t, params):
  a,b,c = params
  return a*(t**2) + b*t + c

params = torch.randn(3).requires_grad_()
params
````

## 1. 코드 요소별 상세 해석## 1) 예측 모델 함수 정의

def f(t, params):
  a,b,c = params
  return a*(t**2) + b*t + c


* def f(t, params):
* 의미: 입력 데이터 t와 모델의 가중치들인 params를 입력받아 예측값을 계산하는 함수 f를 정의합니다. 여기서 t는 앞서 만들었던 시간(time) 데이터에 해당합니다.

* return a*(t**2) + b*t + c
* 의미: 중학교 수학 시간에 배운 2차 함수 공식($y = ax^2 + bx + c$)을 그대로 계산하여 반환합니다.
   * 역할: 우리가 만든 데이터가 2차 곡선 형태를 띠고 있었기 때문에, 모델의 뼈대 역시 2차 함수 형태로 설계한 것입니다. 인공지능이 학습한다는 것은 데이터에 딱 맞는 최적의 a, b, c 숫자 조합을 찾아내는 과정을 뜻합니다.

## 2) 가중치 초기화 및 미분 설정

params = torch.randn(3).requires_grad_()


* torch.randn(3)
* 의미: 파이토치를 사용해 평균 0, 표준편차 1인 무작위 소수(난수) 3개가 들어있는 배열(텐서)을 만듭니다.
   * 역할: 학습을 시작하기 전, 아무것도 모르는 모델의 첫 가중치 [a, b, c]를 완전히 무작위 정답(예: [-0.15, 1.42, -0.89])으로 설정하는 것입니다. 당연히 처음에는 엉터리 예측을 하겠지만, 이 값을 시작으로 경사하강법을 거치며 점점 정답에 수렴하게 됩니다.
* .requires_grad_()
* 의미: 파이토치에게 "이 3개의 숫자(a, b, c)는 앞으로 미분을 통해 값을 계속 수정해 나갈 주인공들이니, 연산 과정을 다 기록해줘"라고 선언하는 명령어입니다.
   * 역할: 이 설정을 켜두어야만 나중에 loss.backward()를 실행했을 때 각 가중치의 기울기(grad)가 정상적으로 계산됩니다.
