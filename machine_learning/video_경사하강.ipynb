 ## 2\. `price = 3*time - 2 + torch.randn(len(time)) * 3`

price = 3*time - 2 + torch.randn(len(time)) * 3

 이 한 줄은 **시간에 따라 증가하는 가격 데이터를 만들되, 현실적인 것처럼 약간의 랜덤한 오차를 넣는 것**입니다.

 조금씩 분해해보겠습니다.

 # 2-1. `3*time`

```
3 * time
```

 앞에서 `time`이:

```
[0, 1, 2, 3, ..., 19]
```

 였죠.

 여기에 각각 3을 곱합니다.

```
3 × [0, 1, 2, 3, ..., 19]
```

 따라서:

```
[0, 3, 6, 9, 12, ..., 57]
```

 이 됩니다.

 수학적으로는:

 $$
3t
$$

 입니다.

---

 # 2-2. `3*time - 2`

 이제 각각의 값에서 2를 뺍니다.

```
3*time - 2
```

 따라서:

```
time = 0 → 3×0 - 2 = -2
time = 1 → 3×1 - 2 =  1
time = 2 → 3×2 - 2 =  4
time = 3 → 3×3 - 2 =  7
...
```

 즉 기본적인 가격은 다음과 같은 직선 형태입니다.

 $$
price = 3t - 2
$$

 여기서 중요한 부분이 있습니다.

 ### `3`은 기울기입니다.

```
price = 3 × time - 2
        ↑
      기울기
```

 시간이 1 증가할 때마다 가격은 평균적으로 **3만큼 증가**합니다.

 ### `-2`는 절편입니다.

```
price = 3 × time - 2
                   ↑
                 절편
```

 시간이 0일 때:

 $$
price = 3(0)-2=-2
$$

 입니다.

 따라서 기본적으로 만들어지는 데이터는 **기울기 3, 절편 -2인 직선**을 따릅니다.

---

 # 2-3. `torch.randn(len(time))`

 이 부분이 조금 중요합니다.

```
torch.randn(len(time))
```

 `randn()`은 **정규분포(normal distribution)를 따르는 랜덤한 숫자**를 만들어냅니다.

 예를 들어 실행할 때마다 대략 이런 값이 나올 수 있습니다.

```
tensor([
    0.21,
   -0.73,
    1.15,
    0.08,
   -1.42,
    ...
])
```

 중요한 특징은 평균이 대략 **0**이고 표준편차가 **1**인 정규분포에서 숫자를 뽑는다는 것입니다.

 즉:

```
torch.randn(...)
```

 은 쉽게 말하면:

 > "평균 0 근처에서 랜덤한 오차를 만들어줘."

 라고 이해하면 됩니다.

---

 ## 그런데 왜 `len(time)`일까요?

```
len(time)
```

 을 먼저 보면:

```
time = [0, 1, 2, ..., 19]
```

 이므로 데이터가 **20개**입니다.

 따라서:

```
len(time)
```

 은

```
20
```

 입니다.

 그러므로:

```
torch.randn(len(time))
```

 은 결국:

```
torch.randn(20)
```

 과 같습니다.

 즉 랜덤 숫자 **20개**를 생성합니다.

 예를 들어:

```
[ 0.4, -0.8, 0.1, 1.3, -0.5, ...]
```

 처럼 됩니다.

---

 # 2-4. `torch.randn(len(time)) * 3`

 이제 랜덤 숫자에 3을 곱합니다.

```
torch.randn(len(time)) * 3
```

 예를 들어 랜덤 숫자가:

```
[0.2, -0.5, 1.0, -0.3]
```

 였다면:

```
[0.6, -1.5, 3.0, -0.9]
```

 가 됩니다.

 이것을 **노이즈(noise), 즉 오차**라고 생각하면 이해하기 쉽습니다.

 # 2-5\. 전체를 합쳐보겠습니다

 다시 원래 코드로 돌아가겠습니다.

```
price = 3*time - 2 + torch.randn(len(time)) * 3
```

 수학적으로 표현하면:

 $$
price = 3t - 2 + noise
$$

 입니다.

 조금 더 정확하게 쓰면:

 $$
price_i = 3t_i - 2 + \epsilon_i
$$

 입니다.

 여기서:

 - $t_i$ = 시간
- $3t_i-2$ = 우리가 의도적으로 만든 기본적인 가격
- $\epsilon_i$ = 랜덤 오차(noise)
- `price_i` = 최종적으로 만들어진 가격

 입니다.

 # 왜 굳이 랜덤 오차를 넣었을까요?

 만약 다음과 같이 했다면:

```

```

 데이터는 정확히 직선 위에 놓입니다.

```
price
  |
55|                 ●
  |              ●
40|           ●
  |        ●
25|     ●
  |  ●
10|
  +---------------------- time
```

 그런데 실제 현실의 데이터는 이렇게 정확하게 직선으로 나오지 않습니다.

 예를 들어 시간이 지나면서 가격이 대체로 올라가더라도:

```
시간 증가 → 가격 증가
```

 하면서도 매번 정확히 3씩 증가하지는 않습니다.

 그래서:

```
+ torch.randn(len(time)) * 3
```

 를 추가하여 **현실적인 데이터처럼 약간의 흔들림을 만들어준 것**입니다.

 ## 이 예제가 머신러닝에서 중요한 이유

 이 데이터는 사실 머신러닝에서 아주 중요한 형태입니다.

 우리가 데이터를 만들 때는 이미 정답을 알고 있습니다.

```
price = 3*time - 2 + noise
```

 즉 실제 데이터가 만들어진 규칙은 대략:

 $$
y = 3x - 2
$$

 입니다.

 그런데 머신러닝에서는 이 **3과 -2를 우리가 직접 알려주지 않고**, 데이터만 보여준 다음 모델이 스스로 찾아내도록 합니다.

 예를 들어 모델이 다음과 같은 식을 학습한다고 생각해 보세요.

 $$
y = ax + b
$$

 처음에는:

```
a = 아무 값
b = 아무 값
```

 일 수 있습니다.

 그런데 데이터를 계속 학습하면서:

```
a ≈ 3
b ≈ -2
```

 에 가까워지도록 만드는 것이 **선형 회귀의 핵심 아이디어**입니다.

 따라서 지금 코드에서 특히 기억해두면 좋은 것은:

```
price = 3*time - 2 + torch.randn(len(time)) * 3
```

 가 단순한 숫자 생성 코드가 아니라,

 > **“정답이 대략 `y = 3x - 2`인 데이터를 만들고, 여기에 랜덤한 오차를 추가해서 실제 데이터처럼 만든다.”**

 라는 의미라는 것입니다.

 그리고 바로 다음 단계에서 이 데이터를 이용해 **`a`와 `b`를 머신러닝이 어떻게 찾아가는지**를 보면 `loss`, `gradient`, `optimizer`, `fit` 같은 개념이 훨씬 쉽게 연결됩니다.

````python
def f(t, params):
    a, b = params
    return a * t * b
````
